```python
# Now I understand the structure - results contain full_rank accuracy
# Let's collect key results from all experiments

def collect_results_from_folder(folder_path):
    results = {}
    if os.path.exists(folder_path):
        for f in os.listdir(folder_path):
            if f.endswith('.json'):
                layer = int(f.replace('.json', ''))
                with open(os.path.join(folder_path, f), 'r') as file:
                    data = json_lib.load(file)
                    if isinstance(data, dict) and 'full_rank' in data:
                        results[layer] = data['full_rank']['accuracy']
                    elif isinstance(data, dict) and 'accuracy' in data:
                        results[layer] = data['accuracy']
    return results

# Collect all results for Meta-Llama-3-70B-Instruct
model_name = "Meta-Llama-3-70B-Instruct"
base_path = os.path.join(results_path, 'causalToM_novis', model_name)

# Answer lookback results
answer_pointer = collect_results_from_folder(os.path.join(base_path, 'answer_lookback', 'pointer'))
answer_payload = collect_results_from_folder(os.path.join(base_path, 'answer_lookback', 'payload'))

# Binding lookback results
binding_addr_payload = collect_results_from_folder(os.path.join(base_path, 'binding_lookback', 'address_and_payload'))
binding_source_1 = collect_results_from_folder(os.path.join(base_path, 'binding_lookback', 'source_1'))
binding_source_2 = collect_results_from_folder(os.path.join(base_path, 'binding_lookback', 'source_2'))

print(f"Answer Lookback Pointer: {len(answer_pointer)} layers")
print(f"  Max IIA at layer {max(answer_pointer, key=answer_pointer.get)}: {max(answer_pointer.values()):.3f}")
print(f"  Layers with IIA > 0.8: {[l for l, v in sorted(answer_pointer.items()) if v > 0.8]}")

print(f"\nAnswer Lookback Payload: {len(answer_payload)} layers")
print(f"  Max IIA at layer {max(answer_payload, key=answer_payload.get)}: {max(answer_payload.values()):.3f}")
print(f"  Layers with IIA > 0.8: {[l for l, v in sorted(answer_payload.items()) if v > 0.8]}")

print(f"\nBinding Address+Payload: {len(binding_addr_payload)} layers")
print(f"  Max IIA at layer {max(binding_addr_payload, key=binding_addr_payload.get)}: {max(binding_addr_payload.values()):.3f}")
print(f"  Layers with IIA > 0.8: {[l for l, v in sorted(binding_addr_payload.items()) if v > 0.8]}")

print(f"\nBinding Source 1: {len(binding_source_1)} layers")
print(f"  Max IIA at layer {max(binding_source_1, key=binding_source_1.get)}: {max(binding_source_1.values()):.3f}")
print(f"  Layers with IIA > 0.8: {[l for l, v in sorted(binding_source_1.items()) if v > 0.8]}")
```
