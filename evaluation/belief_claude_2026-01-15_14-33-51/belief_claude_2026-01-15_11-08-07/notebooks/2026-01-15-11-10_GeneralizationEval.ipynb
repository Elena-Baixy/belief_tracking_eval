## GT1: Generalization to a New Model

### Evidence Examined

**Source Files:**
- `plan.md` (lines 11-12): States methodology uses "Llama-3-70B-Instruct and Llama-3.1-405B-Instruct"
- `CodeWalkthrough.md` (line 3): Confirms testing on both models

- `scripts/evaluate_all_models.py`: Infrastructure to run evaluations across multiple models

### Explicit Arguments for Model Generalization

1. **Empirical Testing on Two Models**: The documentation explicitly states that experiments were conducted on both Llama-3-70B-Instruct and Llama-3.1-405B-Instruct. From `plan.md`:
   > "Analyze Llama-3-70B-Instruct and Llama-3.1-405B-Instruct on 80 correctly-answered samples"

2. **Architectural Reasoning**: The documentation provides mechanistic reasoning for why the lookback mechanism should transfer across transformer models. The method relies on:
   - QK circuits for forming attention patterns (address-pointer mechanism)
   - OV circuits for value transfer (payload mechanism)
   - These are fundamental components shared across transformer-based LLMs

3. **Explicit Condition Stated**: From the existing `doc_only_evaluation/generalization_eval_summary.json`:
   > "mechanism generalizes to other models capable of reliably performing the task"

4. **Infrastructure for Multi-Model Testing**: The `scripts/evaluate_all_models.py` and `src/models.txt` demonstrate that the codebase is designed for cross-model evaluation, with 14 different models listed.

### Assessment

**PASS** — The repository explicitly:
- Tests on two different models (70B and 405B variants)
- Provides architectural reasoning based on shared transformer properties (QK/OV circuits)
- States explicit conditions for when generalization applies (models that can solve the task)
- Includes infrastructure for testing on additional model families

## GT3: Method / Specificity Generalizability

### Does the Work Propose a New Method?

**Yes** — The repository proposes the "lookback mechanism" as a method for understanding how language models track beliefs. This involves:
1. Causal mediation analysis with interchange interventions
2. A hypothesized high-level causal model with ordering IDs (OIs)
3. Desiderata-based Component Masking (DCM) for subspace identification
4. The lookback framework: address/pointer copies enabling retrieval

### Evidence Examined

**Source Files:**
- `plan.md`: Describes the lookback hypothesis and methodology
- `CodeWalkthrough.md`: Overview of the experimental framework
- `notebooks/causalToM_novis/utils.py`: Implementation of experiment generation functions
- `scripts/patching_scripts/`: Interchange intervention implementation

### Arguments For Method Generalizability (or Lack Thereof)

1. **Speculative Claims Without Justification**: The existing documentation expresses "optimism" that the lookback mechanism may apply to "other capabilities in LMs," but this is stated speculatively rather than with explicit justification.

2. **No Applicability Criteria Provided**: The documentation does not specify:
   - What properties a task must have for the lookback framework to be applicable
   - Which aspects of the method are specific to belief tracking vs. general
   - Concrete criteria for when the method would or would not transfer

3. **No Testing Beyond Belief Tracking**: The codebase only applies the method to:
   - Belief tracking (Theory of Mind) tasks
   - Both CausalToM and BigToM are belief tracking datasets
   - No experiments on other task types (e.g., factual recall, arithmetic, reasoning)

4. **Task-Specific Design Elements**: Many components appear belief-tracking-specific:
   - Character/object/state OI assignments
   - Visibility sentence handling
   - Story template structure with belief-inducing scenarios

### Assessment

**FAIL** — The repository:
- Proposes a new method (lookback mechanism framework)
- Expresses optimism about broader applicability but without explicit justification
- Does not provide criteria for when the method would be applicable to other tasks
- Tests only on belief tracking tasks, providing no evidence for generalization beyond this domain
- Contains task-specific design elements without explanation of which are incidental vs. essential

## Summary

### Overall Generalizability Assessment

The repository demonstrates **strong support for generalizability within the belief tracking domain**:

**Strengths:**
- Empirical testing on two model sizes (70B and 405B) with consistent findings
- Clear architectural reasoning for why the mechanism should transfer across transformers
- Testing on both controlled (CausalToM) and external (BigToM) datasets
- Explicit statement of conditions and limitations for generalization
- Infrastructure designed for multi-model evaluation

**Limitations:**
- Method generalizability beyond belief tracking is asserted speculatively without justification
- No testing on task types outside Theory of Mind
- Task-specific design elements may limit transfer of the framework

**Conclusion:**
The paper's core finding—that LMs use lookback mechanisms for belief tracking—is well-supported with evidence for generalization across models and datasets **within the Theory of Mind domain**. However, broader claims about the lookback mechanism as a general pattern for "other capabilities in LMs" are not sufficiently justified by the evidence and arguments present in the repository.
