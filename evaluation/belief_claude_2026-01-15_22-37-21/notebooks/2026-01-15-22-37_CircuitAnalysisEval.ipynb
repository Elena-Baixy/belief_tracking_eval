```python
# Read one of the original JSON files to understand the format
with open(os.path.join(pointer_dir, '46.json'), 'r') as f:
    sample_original = json.load(f)
    
print("Sample original pointer result (layer 46):")
print(json.dumps(sample_original, indent=2)[:2000])
```

```python
# Let's collect the original results for all layers - full_rank accuracy for pointer experiment
original_pointer_iia = {}
for filename in os.listdir(pointer_dir):
    layer = int(filename.replace('.json', ''))
    with open(os.path.join(pointer_dir, filename), 'r') as f:
        data = json.load(f)
        original_pointer_iia[layer] = data['full_rank']['accuracy']
        
# Sort by layer
original_pointer_iia = dict(sorted(original_pointer_iia.items()))
print("Original Pointer IIA by layer (70B model):")
for layer, iia in original_pointer_iia.items():
    print(f"  Layer {layer}: {iia:.3f}")
```

```python
# Check payload results too
payload_dir = os.path.join(answer_lookback_dir, 'payload')
original_payload_iia = {}
for filename in os.listdir(payload_dir):
    layer = int(filename.replace('.json', ''))
    with open(os.path.join(payload_dir, filename), 'r') as f:
        data = json.load(f)
        original_payload_iia[layer] = data['full_rank']['accuracy']
        
original_payload_iia = dict(sorted(original_payload_iia.items()))
print("Original Payload IIA by layer (70B model):")
for layer, iia in original_payload_iia.items():
    print(f"  Layer {layer}: {iia:.3f}")
```

```python
# Check binding_lookback results
binding_dir = os.path.join(llama70b_results, 'binding_lookback')
print("Contents of binding_lookback results:")
for item in os.listdir(binding_dir):
    print(f"  {item}")
```

```python
# Check address_and_payload binding results
binding_addr_dir = os.path.join(binding_dir, 'address_and_payload')
original_binding_iia = {}
for filename in os.listdir(binding_addr_dir):
    layer = int(filename.replace('.json', ''))
    with open(os.path.join(binding_addr_dir, filename), 'r') as f:
        data = json.load(f)
        original_binding_iia[layer] = data['full_rank']['accuracy']
        
original_binding_iia = dict(sorted(original_binding_iia.items()))
print("Original Binding (Address+Payload) IIA by layer (70B model):")
for layer, iia in original_binding_iia.items():
    print(f"  Layer {layer}: {iia:.3f}")
```

```python
# Now let's do a comprehensive comparison between original (70B) and replication (8B)
# We need to compare relative layer positions since model depths differ

print("=" * 80)
print("COMPREHENSIVE COMPARISON: ORIGINAL (70B, 80 layers) vs REPLICATION (8B, 32 layers)")
print("=" * 80)

# Helper function to normalize layer to relative depth
def get_relative_depth(layer, total_layers):
    return layer / total_layers

# Compare pointer experiment
print("\n--- ANSWER LOOKBACK POINTER ---")
print("\nOriginal (70B) key findings:")
print(f"  - Peak IIA: {max(original_pointer_iia.values()):.3f} at layer {max(original_pointer_iia, key=original_pointer_iia.get)}")
print(f"  - Peak relative depth: {max(original_pointer_iia, key=original_pointer_iia.get)/80:.3f}")
high_iia_layers = [l for l, v in original_pointer_iia.items() if v >= 0.9]
print(f"  - Layers with IIA >= 0.9: {min(high_iia_layers)}-{max(high_iia_layers)} (rel depth: {min(high_iia_layers)/80:.3f}-{max(high_iia_layers)/80:.3f})")

print("\nReplication (8B) claims:")
print("  - Peak IIA: 0.83 at layers 16-20")
print("  - Peak relative depth: 16/32 = 0.5 to 20/32 = 0.625")

# The replication claims match layers 34-52/80 = 0.425-0.65 relative depth
# Replication uses 16-20/32 = 0.5-0.625 relative depth
print("\nComparison:")
print(f"  Original high IIA range (>=0.9): layers 34-52 (rel depth 0.425-0.65)")
print(f"  Replication high IIA range: layers 16-20 (rel depth 0.5-0.625)")
print(f"  MATCH: Relative positions are consistent (within ~10%)")
```

```python
# Compare payload experiment
print("\n--- ANSWER LOOKBACK PAYLOAD ---")
print("\nOriginal (70B) key findings:")
print(f"  - Peak IIA: {max(original_payload_iia.values()):.3f}")
high_payload_layers = [l for l, v in original_payload_iia.items() if v >= 0.9]
print(f"  - Layers with IIA >= 0.9: {min(high_payload_layers)}-{max(high_payload_layers)} (rel depth: {min(high_payload_layers)/80:.3f}-{max(high_payload_layers)/80:.3f})")
print(f"  - IIA reaches ~0.8 at layer 56 (rel depth: 0.70)")

print("\nReplication (8B) claims:")
print("  - Peak IIA: 1.00 at layers 30-31")
print("  - IIA reaches 0.88 at layer 28 (rel depth: 0.875)")

print("\nComparison:")
print(f"  Original: IIA rises sharply around layer 56 (rel depth 0.70)")
print(f"  Replication: IIA rises sharply around layer 26-28 (rel depth 0.81-0.875)")
print(f"  Both show payload information emerging in late layers")
print(f"  Relative position difference: ~10-15% but pattern consistent")
```

```python
# Compare binding experiment
print("\n--- BINDING ADDRESS AND PAYLOAD ---")
print("\nOriginal (70B) key findings:")
print(f"  - Peak IIA: {max(original_binding_iia.values()):.3f} at layer {max(original_binding_iia, key=original_binding_iia.get)}")
print(f"  - Peak relative depth: {max(original_binding_iia, key=original_binding_iia.get)/80:.3f}")
high_binding_layers = [l for l, v in original_binding_iia.items() if v >= 0.75]
print(f"  - Layers with IIA >= 0.75: {min(high_binding_layers)}-{max(high_binding_layers)} (rel depth: {min(high_binding_layers)/80:.3f}-{max(high_binding_layers)/80:.3f})")

print("\nReplication (8B) claims:")
print("  - Peak IIA: 0.54 at layer 14")
print("  - Peak relative depth: 14/32 = 0.4375")
print("  - High IIA range: layers 12-16 (rel depth: 0.375-0.5)")

print("\nComparison:")
print(f"  Original: Peak at layers 33-38 (rel depth 0.41-0.475)")
print(f"  Replication: Peak at layers 12-16 (rel depth 0.375-0.5)")
print(f"  MATCH: Relative positions align well (within 5%)")
```
