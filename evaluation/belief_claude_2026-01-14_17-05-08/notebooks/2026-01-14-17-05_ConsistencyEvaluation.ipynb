```python
# Let's read some result files to verify the results
# Read results from Meta-Llama-3-70B-Instruct binding_lookback experiments

import json
import os

model_dir = '/net/scratch2/smallyan/belief_tracking_eval/results/causalToM_novis/Meta-Llama-3-70B-Instruct'

# Read binding_lookback address_and_payload results
addr_payload_dir = os.path.join(model_dir, 'binding_lookback', 'address_and_payload')
addr_payload_results = {}
for f in os.listdir(addr_payload_dir):
    if f.endswith('.json'):
        layer = int(f.replace('.json', ''))
        with open(os.path.join(addr_payload_dir, f), 'r') as fp:
            data = json.load(fp)
            addr_payload_results[layer] = data

# Sort and display
sorted_layers = sorted(addr_payload_results.keys())
print("Binding Address and Payload Results:")
for layer in sorted_layers[:15]:
    data = addr_payload_results[layer]
    print(f"Layer {layer}: IIA = {data.get('iia', data)}")
```

```python
# Let's read the results more cleanly
import json
import os

model_dir = '/net/scratch2/smallyan/belief_tracking_eval/results/causalToM_novis/Meta-Llama-3-70B-Instruct'

# Read binding_lookback address_and_payload results
addr_payload_dir = os.path.join(model_dir, 'binding_lookback', 'address_and_payload')
print("Binding Address and Payload Full Rank IIA Results:")
for layer in sorted([int(f.replace('.json', '')) for f in os.listdir(addr_payload_dir) if f.endswith('.json')])[:10]:
    fpath = os.path.join(addr_payload_dir, f'{layer}.json')
    with open(fpath, 'r') as fp:
        data = json.load(fp)
        full_rank_iia = data.get('full_rank', {}).get('accuracy', 'N/A')
        print(f"Layer {layer}: Full Rank IIA = {full_rank_iia}")
```
