## Repository Understanding Summary

Based on the exploration, the repository studies **belief tracking in language models** with the following key findings:

### Original Experiments:
- **Models used**: Llama-3-70B-Instruct, Llama-3.1-405B-Instruct (primary), with evaluations also on Qwen2.5, OLMo, Gemma models
- **Dataset**: CausalToM dataset with synthetic stories about characters tracking beliefs about objects/containers
- **Method**: Causal mediation analysis with interchange interventions to trace information flow

### Key Neuron-Level Findings:
1. **Answer Lookback Mechanism**: 
   - Answer payload (state token value) localizes to final token residual stream after layer 56
   - Answer pointer information encoded at final token layers 34-52

2. **Binding Lookback Mechanism**:
   - Strongest alignment occurs between layers 33-38 at state token residual stream
   - Source reference (character and object OIs) encoded in layers 20-34

3. **Visibility Lookback Mechanism**:
   - Visibility ID source encoded in layers 10-23
   - Payload aligns after layer 31 at lookback tokens

### For Generalizability Evaluation:
- **GT1 (Model)**: Need to test if layer-specific findings transfer to a new model NOT used in original work
- **GT2 (Data)**: Need to test with new story templates/entities not in the dataset
- **GT3 (Method)**: The method is causal mediation/abstraction - can be applied to other belief tracking or similar tasks

```python
# Create new test stories using the template structure
# We'll test with a small model (Llama-3.1-8B-Instruct) that's available

# Build test prompts following the CausalToM format
instruction = "1. Track the belief of each character as described in the story. 2. A character's belief is formed only when they perform an action themselves or can observe the action taking place. 3. A character does not have any beliefs about the container and its contents which they cannot observe. 4. To answer the question, predict only what is inside the queried container, strictly based on the belief of the character, mentioned in the question. 5. If the queried character has no belief about the container in question, then predict 'unknown'. 6. Do not predict container or character as the final output."

# Template 0: Both characters can't observe each other
def create_test_story(char1, char2, container1, container2, state1, state2):
    story = f"{char1} and {char2} are working in a busy restaurant. To complete an order, {char1} grabs an opaque {container1} and fills it with {state1}. Then {char2} grabs another opaque {container2} and fills it with {state2}. {char2} cannot observe {char1}'s actions. {char1} cannot observe {char2}'s actions."
    return story

# Create 3 test instances
test_instances = [
    {
        "characters": ["Xavier", "Violet"],
        "containers": ["thermos", "carafe"],
        "states": ["lemonade", "smoothie"],
        "questions": [
            ("What does Xavier believe the thermos contains?", "lemonade"),  # Xavier knows
            ("What does Xavier believe the carafe contains?", "unknown"),     # Xavier can't observe
            ("What does Violet believe the thermos contains?", "unknown"),    # Violet can't observe
            ("What does Violet believe the carafe contains?", "smoothie"),    # Violet knows
        ]
    },
    {
        "characters": ["Theodore", "Penelope"],
        "containers": ["decanter", "goblet"],
        "states": ["cider", "mead"],
        "questions": [
            ("What does Theodore believe the decanter contains?", "cider"),
            ("What does Penelope believe the goblet contains?", "mead"),
        ]
    },
    {
        "characters": ["Xavier", "Theodore"],
        "containers": ["thermos", "goblet"],
        "states": ["smoothie", "cider"],
        "questions": [
            ("What does Xavier believe the goblet contains?", "unknown"),
            ("What does Theodore believe the thermos contains?", "unknown"),
        ]
    }
]

print("Created 3 test instances with novel entities")
for i, inst in enumerate(test_instances):
    print(f"\nTest {i+1}: {inst['characters']} with {inst['containers']} containing {inst['states']}")
```

The paper proposes a **new method**: Causal Abstraction with Interchange Interventions for localizing belief tracking mechanisms.

**Key Methodological Components:**
1. **Interchange Intervention Analysis (IIA)** - Swapping activations between counterfactual pairs to measure causal influence
2. **Desiderata-based Component Masking** - Identifying low-rank subspaces encoding specific causal variables
3. **Causal Model Alignment** - Mapping high-level causal variables to neural activations
